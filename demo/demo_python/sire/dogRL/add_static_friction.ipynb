```julia
using LinearAlgebra
using ForwardDiff
using Plots
using Ipopt
using MathOptInterface
const MOI = MathOptInterface
```

```julia
#Boilerplate setup code to interface with IPOPT.

struct ProblemMOI <: MOI.AbstractNLPEvaluator
    n_nlp::Int
    m_nlp::Int
    idx_ineq
    obj_grad::Bool
    con_jac::Bool
    sparsity_jac
    sparsity_hess
    primal_bounds
    constraint_bounds
    hessian_lagrangian::Bool
end

function ProblemMOI(n_nlp,m_nlp;
        idx_ineq=(1:0),
        obj_grad=true,
        con_jac=true,
        sparsity_jac=sparsity_jacobian(n_nlp,m_nlp),
        sparsity_hess=sparsity_hessian(n_nlp,m_nlp),
        primal_bounds=primal_bounds(n_nlp),
        constraint_bounds=constraint_bounds(m_nlp,idx_ineq=idx_ineq),
        hessian_lagrangian=false)

    ProblemMOI(n_nlp,m_nlp,
        idx_ineq,
        obj_grad,
        con_jac,
        sparsity_jac,
        sparsity_hess,
        primal_bounds,
        constraint_bounds,
        hessian_lagrangian)
end

function constraint_bounds(m; idx_ineq=(1:0))
    c_l = zeros(m)

    c_u = zeros(m)
    c_u[idx_ineq] .= Inf
    
    return c_l, c_u
end

function row_col!(row,col,r,c)
    for cc in c
        for rr in r
            push!(row,convert(Int,rr))
            push!(col,convert(Int,cc))
        end
    end
    return row, col
end

function sparsity_jacobian(n,m)

    row = Int64[]
    col = Int64[]

    r = 1:m
    c = 1:n

    row_col!(row,col,r,c)

    return collect(zip(row,col))
end

function sparsity_hessian(n,m)

    row = Int64[]
    col = Int64[]

    r = 1:m
    c = 1:n

    row_col!(row,col,r,c)

    return collect(zip(row,col))
end

function MOI.eval_objective(prob::MOI.AbstractNLPEvaluator, x)
    objective(x)
end

function MOI.eval_objective_gradient(prob::MOI.AbstractNLPEvaluator, grad_f, x)
    ForwardDiff.gradient!(grad_f,objective,x)
    return nothing
end

function MOI.eval_constraint(prob::MOI.AbstractNLPEvaluator,g,x)
    constraint!(g,x)
    return nothing
end

function MOI.eval_constraint_jacobian(prob::MOI.AbstractNLPEvaluator, jac, x)
    ForwardDiff.jacobian!(reshape(jac,prob.m_nlp,prob.n_nlp), constraint!, zeros(prob.m_nlp), x)
    return nothing
end

function MOI.features_available(prob::MOI.AbstractNLPEvaluator)
    return [:Grad, :Jac]
end

MOI.initialize(prob::MOI.AbstractNLPEvaluator, features) = nothing
MOI.jacobian_structure(prob::MOI.AbstractNLPEvaluator) = prob.sparsity_jac

function ipopt_solve(x0,prob::MOI.AbstractNLPEvaluator;
        tol=1.0e-6,c_tol=1.0e-6,max_iter=10000)
    x_l, x_u = prob.primal_bounds
    c_l, c_u = prob.constraint_bounds

    nlp_bounds = MOI.NLPBoundsPair.(c_l,c_u)
    block_data = MOI.NLPBlockData(nlp_bounds,prob,true)

    solver = Ipopt.Optimizer()
    solver.options["max_iter"] = max_iter
    solver.options["tol"] = tol
    solver.options["constr_viol_tol"] = c_tol
    
    #Uncomment the following line to turn off verbose IPOPT output
    # solver.options["print_level"] = 0

    x = MOI.add_variables(solver,prob.n_nlp)

    for i = 1:prob.n_nlp
        MOI.add_constraint(solver, x[i], MOI.LessThan(x_u[i]))
        MOI.add_constraint(solver, x[i], MOI.GreaterThan(x_l[i]))
        MOI.set(solver, MOI.VariablePrimalStart(), x[i], x0[i])
    end

    # Solve the problem
    MOI.set(solver, MOI.NLPBlock(), block_data)
    MOI.set(solver, MOI.ObjectiveSense(), MOI.MIN_SENSE)
    MOI.optimize!(solver)

    # Get the solution
    res = MOI.get(solver, MOI.VariablePrimal(), x)

    return res
end
```

这是一个基于**接触动力学**和**非线性规划（NLP）**的摩擦接触模型。根据代码（特别是 `constraint!` 函数和变量结构），该模型试图求解下一时刻的速度、接触力以及相关乘子，使动力学方程和互补条件同时满足。

你可以在 PPT 中使用以下数学表述：

### **数学模型：全隐式摩擦接触动力学**

#### **1. 决策变量 (Decision Variables)**
对于 $N$ 个接触点，系统的未知量向量 $z$ 包含：
*   **速度**: $v \in \mathbb{R}^{3N}$ （下一时刻的接触点速度）
*   **接触力**: $f \in \mathbb{R}^{3N}$，其中每个接触点 $i$ 的力分解为切向分量 $f_{t,i} \in \mathbb{R}^2$ 和法向分量 $f_{n,i} \in \mathbb{R}$
*   **摩擦乘子**: $\lambda \in \mathbb{R}^N$ （与最大耗散原理相关）
*   **松弛变量**: $s \in \mathbb{R}^N$ （用于放松互补条件）

#### **2. 约束方程 (Constraints)**

优化问题需满足以下物理约束：

**I. 离散动力学方程 (Discrete Dynamics)**
$$
M(v - v_0) = h \cdot (b + J^T f)
$$
或者根据代码中的形式（使用了逆质量矩阵或响应矩阵 $A$）：
$$
v = v_{\text{hist}} + h \cdot b + h \cdot A \cdot f
$$
*   其中 $v_{\text{hist}}$ 是历史速度，$b$ 是外部加速度项（如重力、科氏力），$A$ 是接触空间的逆有效质量矩阵（Delassus Matrix）。

**II. 最大耗散原理 (Maximum Dissipation Principle)**
对于每个接触点 $i$，切向速度方向应与摩擦力方向相反：
$$
\begin{bmatrix} v_{t,i}^x \\ v_{t,i}^y \end{bmatrix} + \lambda_i \frac{f_{t,i}}{\|f_{t,i}\|} = 0
$$
*   这保证了摩擦力做功最大化（即阻碍运动）。
*   注：代码中使用了平滑范数 $\sqrt{x^2 + \epsilon^2}$ 来避免分母为零。

**III. 摩擦锥约束 (Coulomb Friction Cone)**
接触力必须位于摩擦锥内：
$$
\mu f_{n,i} - \|f_{t,i}\| \ge 0
$$
*   $\mu$ 是摩擦系数。

**IV. 松弛互补条件 (Relaxed Complementarity)**
为了处理接触状态的切换（粘滞/滑动），引入松弛变量 $s_i$：
$$
s_i = \lambda_i \cdot (\mu f_{n,i} - \|f_{t,i}\|)
$$
*   物理意义：当处于滑动状态（$\|f_t\| = \mu f_n$）时，滑移率 $\lambda_i > 0$；当处于粘滞状态（$\|f_t\| < \mu f_n$）时，滑移率 $\lambda_i = 0$。

**V. 目标速度/法向约束 (Normal Velocity Constraint)**
$$
v_{n,i} = v_{\text{target}, i}
$$
*   代码中强制指定了法向速度，通常用于模拟非穿透条件或特定的接触行为。

#### **3. 优化目标 (Objective)**
通过最小化松弛变量来尽可能满足互补性：
$$
\min_{v, f, \lambda, s} \sum_{i=1}^{N} s_i
$$
由于 $s_i$ 通常接近于 0，这实际上是在寻找满足精确互补条件的解。

---

### **PPT 展示建议**

将上述公式整理为如下的非线性规划 (NLP) 标准形式：

$$
\begin{aligned}
\text{minimize} \quad & \sum s_i \\
\text{subject to} \quad & v - v_{\text{hist}} - h \cdot b - h \cdot A \cdot f = 0 & (\text{动力学}) \\
& v_{t,i} + \lambda_i \frac{f_{t,i}}{\|f_{t,i}\|_{\epsilon}} = 0 & (\text{最大耗散}) \\
& \mu f_{n,i} - \|f_{t,i}\|_{\epsilon} \ge 0 & (\text{摩擦锥}) \\
& s_i - \lambda_i (\mu f_{n,i} - \|f_{t,i}\|_{\epsilon}) = 0 & (\text{互补松弛}) \\
& v_{n,i} - v_{\text{target},i} = 0 & (\text{法向约束})
\end{aligned}
$$

```julia
#Objective and constraint functions for IPOPT with Friction
N = 4
function smoothsqrt(x)
    ϵ = 1e-6
    return sqrt(x+ϵ*ϵ) - ϵ
end
function objective(z)
    # z的结构是由 constraint! 中的索引决定的，这里直接按照分块求和即可
    # s 位于 z 的最后 N 个元素: z[7*N+1 : 8*N]
    s_start_index = 7 * N + 1
    return sum(z[s_start_index : end]) #Minimize sum of all slacks
end

function constraint!(c,z)
  for i = 1:N
    # 1. 提取变量
    # 速度 v_i (3维): 位于 Block 1
    v_idx = 3*(i-1)
    v = z[v_idx+1 : v_idx+3]
    
    # 力 f_i (3维): 位于 Block 2
    f_idx = 3*N + 3*(i-1)
    ft = z[f_idx+1 : f_idx+2]
    fn = z[f_idx+3]
    
    # 乘子 λ_i (1维): 位于 Block 3
    λ = z[6*N + i]
    
    # 松弛 s_i (1维): 位于 Block 4
    s = z[7*N + i]
    
    # 2. 计算约束并填入 c
    # 注意: 如果 A 是全局耦合矩阵，这里需要注意。假设这里的 A, b, vhist 已经针对第 i 个点切片或是解耦的。
    # 如果 A 是 3x3 块对角矩阵，下面的写法没问题。如果是全矩阵，需传入完整力向量。
    # 假设当前上下文是针对每个点独立计算或者 A 仅是当前点的逆惯量块：
    
    # (2) 最大耗散 (2维) -> 填入 c 的 Block 2 (3N+1 : 6N) 的前两行
    c[f_idx+1 : f_idx+2] = v[1:2] + λ*ft/sqrt(ft'*ft + 1e-12) 
    # c[f_idx+1 : f_idx+2] = v[1:2] + λ*ft 
    
    # (3) 摩擦锥 (1维) -> 填入 c 的 Block 2 的第3行 (即每个力的法向分量位置)
    # 既然已经不再做分母，我们就不需要 smoothsqrt(x) 中避免除零的微小平移，直接用标准二范数
    norm_ft = sqrt(ft'*ft + 1e-12) # 这里保留了一个极小的 1e-12 避免力严格为 0 处导数奇异
    c[f_idx+3] = μ*fn - norm_ft 
    
    # (4) 松弛互补 (1维) -> 填入 c 的 Block 3 (6N+1 : 7N)
    c[6*N + i] = s - λ*(μ*fn - norm_ft)

    # c[7 * N + i] = smoothsqrt(v[1:2]'*v[1:2]) - λ
    
    # (5) 速度约束 (1维) -> 填入 c 的 Block 4 (7N+1 : 8N)
    c[7*N + i] = v[3] - v_target[i] # 这里假设 v_target 是一个长度为 N 的向量，包含每个点的目标法向速度
  end
  # (1) 动力学约束 (3 * N维) -> 填入 c 的 Block 1 (1:3N)
  c[1 : 3*N] = z[1 : 3*N] - vhist[:] - h * b - A * z[3*N+1:6*N] * h
  return nothing
end

# 指定 c 中哪些是不等式约束 (c[i] >= 0)
# 只有摩擦锥约束是 >= 0。
# 在 c 中的位置是每个力块的第3个分量：3N + 3, 3N + 6, ..., 3N + 3N
nonnegative_constraint_indices = [3*N + 3*i for i in 1:N]

function primal_bounds(n_total)
    # n_total 应该是 8 * N
    # 构造变量上下界
    
    # v (3N): -Inf ~ Inf
    v_bounds = fill(-Inf, 3*N)
    
    # f (3N): 
    # ft (2) 是 -Inf ~ Inf
    # fn (1) 是 0 ~ Inf
    # 重复 N 次
    f_bounds_l = Float64[]
    for i in 1:N
        append!(f_bounds_l, [-Inf, -Inf, 0.0]) # ft_x, ft_y, fn
    end
    
    # λ (N): 0 ~ Inf
    lambda_bounds = zeros(N)
    
    # s (N): 0 ~ Inf
    s_bounds = zeros(N)
    
    # 拼接下界
    x_l = [v_bounds; f_bounds_l; lambda_bounds; s_bounds]
    
    # 上界全是 Inf
    x_u = fill(Inf, 8*N)
    
    return x_l, x_u
end
```

```julia
# =========================================================
# 导入所需的库
import Pkg; Pkg.add(["JuMP", "Clarabel"])
using JuMP
using Clarabel
using LinearAlgebra

# 1. ===== 物理参数初始化 (如前) =====
N = 4
h = 0.001
μ = 0.3

v0 = [0.0526404, 0.0110721, 0.0579474, 0.0691226, 1.03524e-05, 0.0583075, 
      0.0612058, -0.0166839, 0.0579826, 0.0562369, 0.0123911, 0.0585853]
b_ext = [25.8184, 35.8036, 144.122, 19.5788, 59.0772, 316.233, 
         27.917, 54.8199, 323.789, 16.0631, 35.4056, 144.615]
v_target = [0.0582191, 0.0582164, 0.0582121, 0.0582143]

# 从你的输入获取 A
A_dense = [
-4.264208 -0.007513 -0.165674 -0.028791  0.011917  0.000612 -0.154848 -0.027276  0.000104  0.124287  0.006857  0.001586 ;
-0.007513 -3.886122 -2.251832  0.011517 -0.008019 -0.001716  0.024421  0.001278 -0.001996 -0.011560 -0.001604 -0.000798 ;
-0.165674 -2.251832 -6.440301  0.004865 -0.003849 -0.001034  0.007146 -0.000370 -0.000950 -0.002020 -0.000762 -0.000349 ;
-0.028791  0.011517  0.004865 -3.279944 -0.086687  0.051068  0.129998  0.021669  0.003730 -0.153854 -0.018874 -0.006894 ;
 0.011917 -0.008019 -0.003849 -0.086687 -2.494906 -1.079557 -0.026896 -0.009045 -0.003082  0.032468  0.000911 -0.000209 ;
 0.000612 -0.001716 -0.001034  0.051068 -1.079557 -7.358521 -0.004721 -0.003125 -0.000929  0.000559 -0.001941 -0.000942 ;
-0.154848  0.024421  0.007146  0.129998 -0.026896 -0.004721 -3.305642  0.102957 -0.110156 -0.042572 -0.012158 -0.005429 ;
-0.027276  0.001278 -0.000370  0.021669 -0.009045 -0.003125  0.102957 -2.519981 -1.103224 -0.013531 -0.007532 -0.003784 ;
 0.000104 -0.001996 -0.000950  0.003730 -0.003082 -0.000929 -0.110156 -1.103224 -7.348570 -0.000821 -0.001674 -0.001022 ;
 0.124287 -0.011560 -0.002020 -0.153854  0.032468  0.000559 -0.042572 -0.013531 -0.000821 -4.286257  0.012754  0.081210 ;
 0.006857 -0.001604 -0.000762 -0.018874  0.000911 -0.001941 -0.012158 -0.007532 -0.001674  0.012754 -3.905390 -2.264855 ;
 0.001586 -0.000798 -0.000349 -0.006894 -0.000209 -0.000942 -0.005429 -0.003784 -0.001022  0.081210 -2.264855 -6.426958]

# ===== 核心修正方案：块拆解与交替SOCP =====
# W = h * A_dense
# P = 0.5 .* (-W .+ (-W)') # 获取精确严格对称，变负为正用于优化
# v_free = v0 + h * b_ext
# q = -v_free

# ================= 适配 C++ 引擎原生矩阵架构的正确解法 =================
W = - h * A_dense
P = 0.5 .* (W .+ W')    # 提取为正定矩阵，此时 P = -hA
v_free = v0 - h * b_ext
q = v_free              # 引擎物理方程的实质等价于：v = q + P * f

# 将矩阵物理分界：提取法向(n) 和 切向(t) 矩阵结构
idx_n = 3:3:(3*N)              # [3, 6, 9, 12]
idx_t = setdiff(1:3N, idx_n)   # [1,2, 4,5, 7,8, 10,11]

P_nn = P[idx_n, idx_n]
P_nt = P[idx_n, idx_t]
P_tt = P[idx_t, idx_t]

q_n = q[idx_n]
q_t = q[idx_t]

# 迭代初始条件
fn_val = zeros(N)
ft_val = zeros(2*N)

max_iters = 20
println("\n[Clarabel-Iterative] 融合引擎底层符号求解开始...");
for iter in 1:max_iters
    # ---- 1. 法向量支持力求解 ----
    # 物理要求: v_n = q_n + P_nn*f_n + P_nt*f_t = v_target
    # 移项得: P_nn * f_n = v_target - q_n - P_nt*f_t
    fn_new = P_nn \ (v_target - q_n - P_nt * ft_val)
    fn_new = max.(fn_new, 0.0) # 这里的 f_n > 0 算出的是引擎所需的反作用力标量
    fn_val = fn_new
    
    # ---- 2. 给定边界 mu*(fn_val)，分离出切向受限阻尼问题 ----
    model = Model(Clarabel.Optimizer)
    set_silent(model)
    
    @variable(model, ft[1:2N])
    
    # 顺应底层方程体系：使 ∇E(ft) 引导出引擎能接受的同向摩擦力
    # q_tt_current = P_nt' * fn_val - q_t
    # @objective(model, Min, 0.5 * ft' * P_tt * ft + q_tt_current' * ft)
    q_tt_current = P_nt' * fn_val + q_t      # 注意：这里的 - q_t 变成了 + q_t
    @objective(model, Min, 0.5 * ft' * P_tt * ft + q_tt_current' * ft)
    # 二阶摩擦锥严格约束: 法向力是正数
    for i in 1:N
        ft_idx = 2*i-1 : 2*i
        f_bound = μ * fn_val[i]
        @constraint(model, [f_bound; ft[ft_idx]] in SecondOrderCone())
    end
    
    optimize!(model)
    ft_new = value.(ft)
    
    # 收敛检查评估
    err = norm(ft_new - ft_val)
    ft_val = ft_new
    if err < 1e-6
        println("[Clarabel-Iterative] 迭代已在第 $iter 次内精准收敛!")
        break
    end
end

# ===== 结果组装验证 =====
f_opt = zeros(3N)
f_opt[idx_n] = fn_val
f_opt[idx_t] = ft_val

# 计算最终速度验证
v_opt = q + P * f_opt # 严格等价于 C++ 内部的 v0 + h*b_ext + h*A*f

println("\n========== Clarabel (原生动力学匹配体系) 最优解 ==========")
for i in 1:N
    idx = 3*(i-1)
    println("Contact Point $i:")
    println("  Vel  (v0) = ", round.(v0[idx+1:idx+3], digits=11))
    println("  Vel   (v) = ", round.(v_opt[idx+1:idx+3], digits=11))
    println("  Force (f) = ", round.(f_opt[idx+1:idx+3], digits=6))
end
```
