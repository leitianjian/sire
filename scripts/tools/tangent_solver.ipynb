```python
import numpy as np
from scipy.optimize import nnls
from scipy import linalg

def compareNNLSvsSolve(A, b):
  A = np.array(A).reshape((4,4))
  b = np.array(b).reshape((4,1))
  f_opt, residual = nnls(A, b)  # 非负最小二乘
  print("NNLS:", f_opt, residual)
  try:
    x_scipy = linalg.solve(A, b)
    residual = np.linalg.norm(b - A @ x_scipy)
    print("solve: ", x_scipy, residual)
  except np.linalg.LinAlgError:
    print("矩阵A奇异，无法求解精确解。")

A = [-4.74963, -0.00191284, 0.00754722, 0.00831402, 0.00125272, -3.71874, -0.000425759, 0.000148162, -0.00290603, 0.00651581, -3.64783, -0.00488327, 0.000694607, -0.000123184, -0.000415422, -5.46154]  # 维度 n×n
b = [255.341, -369.199, -503.518, 266.192]
compareNNLSvsSolve(A, b)
```
